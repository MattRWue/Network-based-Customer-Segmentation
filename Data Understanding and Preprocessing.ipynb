Preprocessing

```python
# Combine prior and train order-product datasets
order_products = pd.concat([prior, train],axis=0,ignore_index=True)
print(order_products.shape)
order_products.head()
```

```python
# Merge with orders table
preprocessed_df = order_products.merge(orders,on="order_id",how="left")
print(preprocessed_df.shape)
preprocessed_df.head()
```

```python
# Merge with products table
preprocessed_df = preprocessed_df.merge(products,on="product_id",how="left")

# Merge with aisles table
preprocessed_df = preprocessed_df.merge(aisles,on="aisle_id",how="left")

# Merge with departments table
preprocessed_df = preprocessed_df.merge(departments,on="department_id",how="left")

print("Final preprocessed dataset:")
print(preprocessed_df.shape)
preprocessed_df.head()
```

Cohort Analysis:

```python
#First Order is Day 0
orders["days_since_prior_order"] = orders["days_since_prior_order"].fillna(0)
#sort by user id and order number
orders = orders.sort_values(["user_id", "order_number"])
#construct cumulative timeline
orders["cumulative_timeline_day"] = orders.groupby("user_id")["days_since_prior_order"].cumsum()
orders.head(100)
```

```python
#Orders within the first 7, 14, 21 and 30 days
orders_7_days = orders[orders["cumulative_timeline_day"] <= 7]
orders_14_days = orders[orders["cumulative_timeline_day"] <= 14]
orders_21_days = orders[orders["cumulative_timeline_day"] <= 21]
orders_30_days = orders[orders["cumulative_timeline_day"] <= 30]

#Average number of orders per customer in each window
avg_orders_7 = orders_7_days.groupby("user_id")["order_id"].count().mean()
avg_orders_14 = orders_14_days.groupby("user_id")["order_id"].count().mean()
avg_orders_21 = orders_21_days.groupby("user_id")["order_id"].count().mean()
avg_orders_30 = orders_30_days.groupby("user_id")["order_id"].count().mean()

#Share of customers with >= two orders in each window
total_customers = orders["user_id"].nunique()
repeat_share_7 = orders_7_days.groupby("user_id")["order_number"].max().ge(2).sum()/ total_customers * 100
repeat_share_14 = orders_14_days.groupby("user_id")["order_number"].max().ge(2).sum()/ total_customers * 100
repeat_share_21 = orders_21_days.groupby("user_id")["order_number"].max().ge(2).sum()/ total_customers * 100
repeat_share_30 = orders_30_days.groupby("user_id")["order_number"].max().ge(2).sum()/ total_customers * 100

#Summarize results
window_summary = pd.DataFrame({
    "window_days": [7, 14, 21, 30],
    "avg_orders_per_customer": [avg_orders_7, avg_orders_14, avg_orders_21, avg_orders_30],
    "repeat_customer_share": [repeat_share_7, repeat_share_14, repeat_share_21, repeat_share_30]
})

window_summary
```

```python
#Frequency of order counts within the first 14 days
early_orders = orders[orders["cumulative_timeline_day"] <= 14]

early_order_counts = early_orders.groupby("user_id")["order_id"].count().reset_index(name="orders_first_14_days")

early_order_counts["orders_first_14_days"].value_counts().sort_index()
```

```python
#cohort assignment
early_order_counts["cohort"] = None

early_order_counts.loc[early_order_counts["orders_first_14_days"] == 1, "cohort"] = "Low Activity"
early_order_counts.loc[early_order_counts["orders_first_14_days"].between(2, 3),"cohort"] = "Medium Activity"
early_order_counts.loc[early_order_counts["orders_first_14_days"] >= 4,"cohort"] = "High Activity"

print(early_order_counts["cohort"].value_counts())
```

```python
print(early_order_counts.shape)
print(early_order_counts.head(10))
```

```python
# Calculate total number of orders per customer
total_orders_per_customer = orders.groupby("user_id")["order_number"].max().reset_index(name="total_orders")

# Add total order count to cohort table
cohort_retention_df = early_order_counts.merge(total_orders_per_customer, on="user_id", how="left")
cohort_retention_df.head()
```

```python
# Calculate retention rates for each cohort and milestone
retention_results = []

for cohort in cohort_retention_df["cohort"].unique():
    
    cohort_data = cohort_retention_df[cohort_retention_df["cohort"] == cohort]
    
    cohort_size = len(cohort_data)

    retained_5 = (cohort_data["total_orders"] >= 5).sum()
    retained_10 = (cohort_data["total_orders"] >= 10).sum()
    retained_15 = (cohort_data["total_orders"] >= 15).sum()
    retained_20 = (cohort_data["total_orders"] >= 20).sum()

    retention_results.append({
        "cohort": cohort,
        "milestone": 5,
        "cohort_size": cohort_size,
        "retained_customers": retained_5,
        "retention_rate": retained_5 / cohort_size * 100
    })

    retention_results.append({
        "cohort": cohort,
        "milestone": 10,
        "cohort_size": cohort_size,
        "retained_customers": retained_10,
        "retention_rate": retained_10 / cohort_size * 100
    })

    retention_results.append({
        "cohort": cohort,
        "milestone": 15,
        "cohort_size": cohort_size,
        "retained_customers": retained_15,
        "retention_rate": retained_15 / cohort_size * 100
    })

    retention_results.append({
        "cohort": cohort,
        "milestone": 20,
        "cohort_size": cohort_size,
        "retained_customers": retained_20,
        "retention_rate": retained_20 / cohort_size * 100
    })

retention_df = pd.DataFrame(retention_results)

retention_df.head(10)
```
